```python
# =========================================================
# 1. IMPORT LIBRARIES
# =========================================================
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import ResNet50
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten, Dropout, BatchNormalization
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint

from sklearn.utils.class_weight import compute_class_weight

# =========================================================
# 2. DATASET PATHS
# =========================================================
train_path = r"D:\ai project medical\chest_xray\train"
val_path = r"D:\ai project medical\chest_xray\val"
test_path = r"D:\ai project medical\chest_xray\test"

# =========================================================
# 3. IMAGE PREPROCESSING
# =========================================================
# Slight adjustment to augmentation to keep X-ray structures realistic
train_gen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=15,
    zoom_range=0.15,
    horizontal_flip=True,
    width_shift_range=0.1,
    height_shift_range=0.1,
    shear_range=0.1
)

test_gen = ImageDataGenerator(
    rescale=1./255
)

# =========================================================
# 4. LOAD TRAIN DATA
# =========================================================
train_data = train_gen.flow_from_directory(
    train_path,
    target_size=(224, 224),
    batch_size=32,  # Increased batch size slightly for more stable gradients
    class_mode="binary",
    shuffle=True
)

# =========================================================
# 5. LOAD VALIDATION DATA
# =========================================================
val_data = test_gen.flow_from_directory(
    val_path,
    target_size=(224, 224),
    batch_size=16,  # Matches the small 16-image validation folder size
    class_mode="binary",
    shuffle=False
)

# =========================================================
# 6. LOAD TEST DATA
# =========================================================
test_data = test_gen.flow_from_directory(
    test_path,
    target_size=(224, 224),
    batch_size=32,
    class_mode="binary",
    shuffle=False
)

# =========================================================
# 7. CLASS WEIGHT BALANCING
# =========================================================
class_weights = compute_class_weight(
    class_weight="balanced",
    classes=np.unique(train_data.classes),
    y=train_data.classes
)
class_weights = dict(enumerate(class_weights))
print("\nClass Weights:\n", class_weights)

# =========================================================
# 8. LOAD RESNET50 MODEL
# =========================================================
base_model = ResNet50(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3),
    pooling='avg'
)

# =========================================================
# 9. FREEZE LAYERS CONSERVATIVELY
# =========================================================
# Unfreezing fewer layers at first helps prevent "representation disruption" 
# in deep networks like ResNet on medical images.
for layer in base_model.layers[:-15]:
    layer.trainable = False
for layer in base_model.layers[-15:]:
    layer.trainable = True

# =========================================================
# 10. CREATE FINAL MODEL WITH BATCH NORMALIZATION
# =========================================================
model = Sequential([
    base_model,
    Dense(256, activation="relu"),
    BatchNormalization(),  # Added to stabilize training updates
    Dropout(0.5),
    Dense(128, activation="relu"),
    BatchNormalization(),
    Dropout(0.3),
    Dense(1, activation="sigmoid")
])

# =========================================================
# 11. COMPILE MODEL
# =========================================================
model.compile(
    optimizer=Adam(learning_rate=1e-5),  # Kept small for stable fine-tuning
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

model.summary()

# =========================================================
# 13. CALLBACKS (WITH BEST WEIGHT CHECKPOINTING)
# =========================================================
checkpoint_path = r"D:\ai project medical\xray_model_best.h5"

checkpoint = ModelCheckpoint(
    filepath=checkpoint_path,
    monitor='val_loss',
    save_best_only=True,   # Keeps ONLY the absolute best epoch weights
    mode='min',
    verbose=1
)

early_stop = EarlyStopping(
    monitor='val_loss',
    patience=6,            # Slightly increased patience to allow the scheduler to work
    restore_best_weights=True
)

reduce_lr = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.2,
    patience=2,
    min_lr=1e-7,
    verbose=1
)

# =========================================================
# 14. TRAIN MODEL
# =========================================================
history = model.fit(
    train_data,
    validation_data=val_data,
    epochs=25,
    class_weight=class_weights,
    callbacks=[early_stop, reduce_lr, checkpoint]
)

# =========================================================
# 15. EVALUATE BEST MODEL ON TEST DATASET
# =========================================================
# Load the actual best saved file to make sure evaluations are accurate
best_model = tf.keras.models.load_model(checkpoint_path)
test_loss, test_accuracy = best_model.evaluate(test_data)

print("\nTest Accuracy of Best Model:", round(test_accuracy * 100, 2), "%")

# =========================================================
# 17. ACCURACY GRAPH
# =========================================================
plt.figure(figsize=(10, 5))
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.title("Model Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend(["Train", "Validation"])
plt.grid()
plt.show()

# =========================================================
# 18. LOSS GRAPH
# =========================================================
plt.figure(figsize=(10, 5))
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.title("Model Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend(["Train", "Validation"])
plt.grid()
plt.show()

# =========================================================
# 19. PREDICT SINGLE IMAGE USING BEST MODEL
# =========================================================
from tensorflow.keras.preprocessing import image

img_path = r"D:\ai project medical\chest_xray\test\PNEUMONIA\person1_virus_6.jpeg"
img = image.load_img(img_path, target_size=(224, 224))

img_array = image.img_to_array(img)
img_array = img_array / 255.0
img_array = np.expand_dims(img_array, axis=0)

prediction = best_model.predict(img_array)
confidence = prediction[0][0]

if confidence > 0.5:
    result = "PNEUMONIA"
    final_confidence = confidence * 100
else:
    result = "NORMAL"
    final_confidence = (1 - confidence) * 100

print("\nPrediction:", result)
print("Confidence:", round(final_confidence, 2), "%")

# =========================================================
# 20. SHOW IMAGE
# =========================================================
plt.figure(figsize=(6, 6))
plt.imshow(img)
plt.title(f"{result} - {round(final_confidence, 2)}%")
plt.axis("off")
plt.show()
```

```python
# =========================================================
# DATASET DISTRIBUTION PIE CHARTS
# =========================================================
import matplotlib.pyplot as plt
import numpy as np

# 1. Extract counts and labels from the data generators
train_labels, train_counts = np.unique(train_data.classes, return_counts=True)
val_labels, val_counts = np.unique(val_data.classes, return_counts=True)
test_labels, test_counts = np.unique(test_data.classes, return_counts=True)

# Get class names (e.g., ['NORMAL', 'PNEUMONIA'])
class_names = list(train_data.class_indices.keys())

# Define matching colors (Soft Blue for Normal, Soft Red/Orange for Pneumonia)
colors = ['#5dade2', '#ec7063']
explode = (0.05, 0) # slightly separate the first slice for visual clarity

# 2. Set up the plotting grid (1 row, 3 columns)
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

# Train Chart
axes[0].pie(train_counts, labels=class_names, autopct='%1.1f%%', 
            startangle=90, colors=colors, explode=explode, shadow=True,
            textprops={'fontsize': 12, 'weight': 'bold'})
axes[0].set_title(f"Train Set\nTotal: {sum(train_counts)} images", fontsize=14, weight='bold')

# Validation Chart
axes[1].pie(val_counts, labels=class_names, autopct='%1.1f%%', 
            startangle=90, colors=colors, explode=explode, shadow=True,
            textprops={'fontsize': 12, 'weight': 'bold'})
axes[1].set_title(f"Validation Set\nTotal: {sum(val_counts)} images", fontsize=14, weight='bold')

# Test Chart
axes[2].pie(test_counts, labels=class_names, autopct='%1.1f%%', 
            startangle=90, colors=colors, explode=explode, shadow=True,
            textprops={'fontsize': 12, 'weight': 'bold'})
axes[2].set_title(f"Test Set\nTotal: {sum(test_counts)} images", fontsize=14, weight='bold')

# Adjust layout and show
plt.tight_layout()
plt.show()
```

```python
# =========================================================
# 21. ACCURACY COMPARISON BAR CHART & CONFUSION MATRIX
# =========================================================
import seaborn as sns
from sklearn.metrics import confusion_matrix, classification_report

# --- CHART 1: ACCURACY COMPARISON BAR CHART ---
# Extract final accuracies from the training history and test evaluation
final_train_acc = history.history['accuracy'][-1] * 100
final_val_acc = history.history['val_accuracy'][-1] * 100
final_test_acc = test_accuracy * 100

categories = ['Train Accuracy', 'Validation Accuracy', 'Test Accuracy']
accuracy_values = [final_train_acc, final_val_acc, final_test_acc]

plt.figure(figsize=(8, 5))
bars = plt.bar(categories, accuracy_values, color=['#3498db', '#e67e22', '#2ecc71'], width=0.5, edgecolor='black', zorder=2)

# Add percentage values on top of each bar
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height + 1,
             f'{round(height, 2)}%',
             ha='center', va='bottom', fontsize=11, weight='bold')

plt.title("Overall Model Accuracy Comparison", fontsize=14, weight='bold')
plt.ylabel("Accuracy (%)", fontsize=12)
plt.ylim(0, 110) # Give extra room on top for text labels
plt.grid(axis='y', linestyle='--', alpha=0.7, zorder=1)
plt.tight_layout()
plt.show()


# --- CHART 2: CONFUSION MATRIX (TEST SET BREAKDOWN) ---
# Reset test generator to ensure deterministic data ordering
test_data.reset()

# Get predictions from the best model
predictions = best_model.predict(test_data, verbose=1)
predicted_classes = (predictions > 0.5).astype(int).reshape(-1)
true_classes = test_data.classes
class_labels = list(test_data.class_indices.keys())

# Generate confusion matrix data
cm = confusion_matrix(true_classes, predicted_classes)

plt.figure(figsize=(6, 5))
# Using Seaborn to create a beautiful heatmap breakdown
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=class_labels, yticklabels=class_labels,
            cbar=False, annot_kws={"size": 14, "weight": "bold"})

plt.title("Confusion Matrix\n(Where did the model get things right/wrong?)", fontsize=13, weight='bold')
plt.xlabel("Predicted Category", fontsize=11)
plt.ylabel("Actual True Category", fontsize=11)
plt.tight_layout()
plt.show()

# Print text-based classification metrics (Precision, Recall, F1-score)
print("\n" + "="*50)
print("DETAILED ACCURACY REPORT")
print("="*50)
print(classification_report(true_classes, predicted_classes, target_names=class_labels))
print("="*50)

```
